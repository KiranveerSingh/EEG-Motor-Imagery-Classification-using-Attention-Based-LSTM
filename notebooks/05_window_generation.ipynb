```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
segments = np.load(
    "../processed/S001_R04_segments.npy"
)

labels = np.load(
    "../processed/S001_R04_labels.npy"
)

print(segments.shape)
print(labels.shape)
```

```python
window_size = 80
step_size = 40
```

```python
all_windows = []

for segment in segments:

    segment_windows = []

    for start in range(
        0,
        320 - window_size + 1,
        step_size
    ):

        end = start + window_size

        window = segment[:, start:end]

        segment_windows.append(window)

    all_windows.append(segment_windows)

all_windows = np.array(all_windows)
```

```python
print(all_windows.shape)
```

```python
print(len(all_windows[0]))
```

```python
fig, axes = plt.subplots(
    7,
    1,
    figsize=(12,10)
)

for i in range(7):

    axes[i].plot(
        all_windows[0, i, 0, :]
    )

    axes[i].set_title(
        f"Window {i+1}"
    )

plt.tight_layout()
plt.show()
```

```python
print("Window Shape:")
print(all_windows[0,0].shape)

print("\nFirst Window:")
print(all_windows[0,0,0,:5])

print("\nSecond Window:")
print(all_windows[0,1,0,:5])
```

```python
np.save(
    "../processed/S001_R04_windows.npy",
    all_windows
)

print("Saved successfully")
```
