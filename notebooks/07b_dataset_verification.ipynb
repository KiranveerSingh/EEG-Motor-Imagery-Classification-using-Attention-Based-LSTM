```python
import numpy as np
```

```python
X = np.load("../processed/X.npy")
y = np.load("../processed/y.npy")

print(X.shape)
print(y.shape)
```

```python
print("Datatype:", X.dtype)

print("NaN values:", np.isnan(X).sum())

print("Inf values:", np.isinf(X).sum())

print("Min:", np.min(X))

print("Max:", np.max(X))

print("Mean:", np.mean(X))

print("Std:", np.std(X))
```

```python
print("Unique labels:")

print(np.unique(y))

print("Class counts:")

print(np.unique(y, return_counts=True))
```

```python
print("Sample shape:")

print(X[0].shape)

print("\nOne timestep shape:")

print(X[0,0].shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,4))

plt.plot(X[0,0])

plt.title("Feature Vector of First Window")

plt.xlabel("Feature Index")

plt.ylabel("Value")

plt.grid()

plt.show()
```
