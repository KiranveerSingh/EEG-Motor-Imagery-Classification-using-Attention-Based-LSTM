```python
pairs = [
    ('FC5','FC6'),
    ('FC3','FC4'),
    ('FC1','FC2'),

    ('C5','C6'),
    ('C3','C4'),
    ('C1','C2'),

    ('CP5','CP6'),
    ('CP3','CP4'),
    ('CP1','CP2'),

    ('FP1','FP2'),

    ('AF7','AF8'),
    ('AF3','AF4'),

    ('F7','F8'),
    ('F5','F6'),
    ('F3','F4'),
    ('F1','F2'),

    ('FT7','FT8'),

    ('T7','T8'),
    ('T9','T10'),

    ('TP7','TP8'),

    ('P7','P8'),
    ('P5','P6'),
    ('P3','P4'),
    ('P1','P2'),

    ('PO7','PO8'),
    ('PO3','PO4'),

    ('O1','O2')
]

print("Number of pairs =", len(pairs))
```

```python
def create_differential_channels(raw, pairs):

    # Clean channel names
    clean_names = {
        ch: ch.replace(".", "").upper()
        for ch in raw.ch_names
    }

    raw.rename_channels(clean_names)

    # Get EEG data
    data = raw.get_data()

    # Create fresh channel index dictionary
    ch_idx = {
        ch: i
        for i, ch in enumerate(raw.ch_names)
    }

    diff_data = []

    for left, right in pairs:

        left_idx = ch_idx[left]
        right_idx = ch_idx[right]

        diff_signal = data[left_idx] - data[right_idx]

        diff_data.append(diff_signal)

    diff_data = np.array(diff_data)

    diff_names = [
        f"{left}-{right}"
        for left, right in pairs
    ]

    return diff_data, diff_names
```

```python
# Test on S001R04

edf_path = "../data/eegmmidb/files/S001/S001R04.edf"

raw = mne.io.read_raw_edf(
    edf_path,
    preload=True,
    verbose=False
)

diff_data, diff_names = create_differential_channels(
    raw,
    pairs
)

print("Shape:", diff_data.shape)
print("First 5 differential channels:")
print(diff_names[:5])
```

```python
print("Min:", np.min(diff_data))
print("Max:", np.max(diff_data))
print("Mean:", np.mean(diff_data))
print("Std:", np.std(diff_data))
```

```python
for run in ["R04", "R08", "R12"]:

    path = f"../data/eegmmidb/files/S001/S001{run}.edf"

    raw = mne.io.read_raw_edf(
        path,
        preload=True,
        verbose=False
    )

    diff_data, diff_names = create_differential_channels(
        raw,
        pairs
    )

    print(run, diff_data.shape)
```

```python
# Re-load R04

raw = mne.io.read_raw_edf(
    "../data/eegmmidb/files/S001/S001R04.edf",
    preload=True,
    verbose=False
)

diff_data, diff_names = create_differential_channels(
    raw,
    pairs
)

np.save(
    "../processed/S001_R04_diff.npy",
    diff_data
)

print("File saved successfully!")
```
